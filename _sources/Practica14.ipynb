## Práctica 14: Creación de interfaces gráficas

Una interfaz gráfica es un elemento muy común en la actualidad, una interfaz gráfica es en esencia un programa que permite la interacción entre un usuario final y un conjunto de sub-programas o código de propósito específico. El objetivo principal de una interfaz gráfica es ofrecer al usuario final de una herramienta computacional basada en código, una manera interactiva, sencilla y en un lenguaje que que no necesariamente exija de dicho usuario un conocimiento especializado o avanzado del lenguaje o lenguajes de programación en el cual se desarrolló.

El desarrollo de una interfaz gráfica es un proceso que involucra creatividad y al mismo tiempo orden y un plan previo para los elementos que esta incluirá y las funciones que contendrá y sus alcances. Por lo tanto, es importante tener en mente y bien delimitado el contenido y la apariencia final que tendrá. Para abrir el ambiente de creación de interfaz gráfica en MATLAB se usa la instrucción:


````matlab
appdesigner
````
